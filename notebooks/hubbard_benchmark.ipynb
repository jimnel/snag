```python
import snag
import time
import matplotlib.pyplot as plt
import numpy as np
```

```python
n_atom = 10
filling = 0.5
n_part = int(filling * n_atom)
t_mat = np.ones((n_atom, n_atom))

print(f"n atoms: {n_atom}\nn part: {n_part}")
results = []

for i in range(n_part + 1):
    t0 = time.time()
    calc = snag.Hubbard(n_part - i, i, n_atom)
    basis = calc.basis
    doubles = calc.doubles()
    kinetic = calc.hopping(t_mat)
    t1 = time.time()
    results.append([n_part - i, i, calc.dim, doubles.sum(), t1 - t0])
    print(" ", *results[-1])

results = np.array(results)
```

```python
# plt.plot(results[:, 1], results[:, 2], marker='o', label='dim')
plt.plot(results[:, 1], results[:, 4], marker="o")

plt.xlabel("n dn")
plt.ylabel("Time (sec)")
```
