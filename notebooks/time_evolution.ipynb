```python
import snag
import qse
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_sites = 4
qbits = qse.lattices.chain(1.0, n_sites)
qbits.draw(radius="nearest")
```

```python
t_mat = -qbits.get_adjacency_matrix(1.0)
plt.imshow(t_mat)
plt.colorbar()
plt.show()
```

```python
calc = snag.Hubbard(n_up=2, n_dn=0, n_sites=qbits.nqbits)
print(calc.basis)

h_t = calc.hopping(t_mat)
h_u = calc.doubles()

hamiltonian = h_t + 1.0 * np.diag(h_u)

initial_state = np.zeros(calc.dim)
initial_state[3] = 1.0
print(initial_state)
```

```python
times, results = qse.calc.evolve(hamiltonian, 1.0, 10, initial_state=initial_state)
print(results)
```

```python
probs = (np.conj(results) * results).real

for i in range(calc.dim):
    plt.plot(times, probs[:, i], label=i)

plt.legend()
plt.show()
```
